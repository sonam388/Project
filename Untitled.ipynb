```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import joblib

data = {
    "Study_hr" : [1,2,3,4,5],
    "Marks":[35,45,55,65,75]
}

df = pd.DataFrame(data)

x = df[['Study_hr']]
y = df['Marks']

model = LinearRegression()

model.fit(x,y)

joblib.dump(model , "model.joblib")
```
